# Part I: Implicit Function Differentiation

**Setup.** The model $M(X_I)$ has internal parameters $X_I \in \mathbb{R}^n$. We observe $n$ market quantities $X_M \in \mathbb{R}^n$, and each model function reprices one of them: $V_j(X_I) = X_M^j$. Since $\dim(X_I) = \dim(X_M)$, this is a **square nonlinear system** of $n$ equations in $n$ unknowns.

---

## Q1. General recipe for solving $X_I$ from $V(X_I) = X_M$

Rewrite calibration as root finding on the residual

$$F(X_I) = V(X_I) - X_M = 0 .$$

**Multi-dimensional Newton–Raphson:**

1. **Initial guess** $X_I^{(0)}$: a flat curve / flat vol, a simple bootstrap, or the previous day's calibration.
2. **Residual:** $F^{(k)} = V(X_I^{(k)}) - X_M$.
3. **Jacobian:** $J^{(k)} = \dfrac{\partial V}{\partial X_I}\Big|_{X_I^{(k)}}$ ($n \times n$), analytically / by AD if possible, otherwise by finite differences (bump each parameter).
4. **Newton step:** solve the linear system $J^{(k)}\,\Delta^{(k)} = -F^{(k)}$ (LU / `np.linalg.solve`; never form $J^{-1}$ explicitly).
5. **Update:** $X_I^{(k+1)} = X_I^{(k)} + \Delta^{(k)}$.
6. **Stop** when $\|F^{(k)}\|_\infty < \varepsilon$ (e.g. $10^{-10}$ in price units) and the step is small.

**Practical points**
- **Robustness:** use damping / line search, or Levenberg–Marquardt, when the initial guess is far from the solution. Equivalently, minimise $\tfrac12\|F\|^2$; for a square, well-posed system the minimum is an exact fit ($F = 0$).
- **Constraints:** reparametrise to keep parameters valid (e.g. $\sigma = e^{y}$ to keep vols positive).
- **Bootstrapping:** if the system is triangular (instrument $j$ depends only on parameters $1,\dots,j$, as in curve building), solve it sequentially with $n$ one-dimensional root finds.
- **Well-posedness:** a locally unique solution requires $J$ to be non-singular near the root, i.e. each market quote must carry independent information about the parameters.

---

## Q2. Computing $\dfrac{dX_I}{dX_M}$ at the optimal $X_I$

At the calibrated solution $X_I^* = X_I^*(X_M)$, the calibration equations hold identically:

$$V\big(X_I^*(X_M)\big) \equiv X_M .$$

Differentiating both sides with respect to $X_M$ (chain rule) gives

$$\underbrace{\frac{\partial V}{\partial X_I}\Big|_{X_I^*}}_{J}\;\frac{dX_I}{dX_M} = I
\quad\Longrightarrow\quad
\boxed{\;\frac{dX_I}{dX_M} = J^{-1}\Big|_{X_I^*}\;}$$

This is the **implicit function theorem**: for $F(X_I, X_M) = V(X_I) - X_M = 0$,
$\dfrac{dX_I}{dX_M} = -\left(\dfrac{\partial F}{\partial X_I}\right)^{-1}\dfrac{\partial F}{\partial X_M} = -J^{-1}(-I) = J^{-1}$.
It is valid whenever $J$ is invertible at $X_I^*$.

**Why it is useful**
- **No re-calibration.** Bump-and-recalibrate needs $n+1$ full calibrations. Here we need only the Jacobian at the solution, which Newton has already computed in its final iteration, plus one linear solve.
- **Risk to market quotes.** A pricer naturally gives risk to model parameters, $\partial PV/\partial X_I$. Traders hedge with market instruments, so we map it with the chain rule:

$$\frac{\partial PV}{\partial X_M} = \frac{\partial PV}{\partial X_I}\, J^{-1}.$$

  (In the HW5 code, risk is computed to the IFR states. Because the IFRs are input directly, the state-to-quote map is the identity. If the curve were calibrated from, e.g., swap rates, $J^{-1}$ is exactly the step that converts IFR risk into swap-rate risk.)

---

## Q3. Calibrating a 3-strike volatility interpolator

**Mapping to the framework**
- Market observables: $X_M = (P_1, P_2, P_3)$, the un-discounted OTM prices at $K_1 < K_{atm} < K_3$ (a put at $K_1$, a call at $K_3$, either at $K_{atm}$).
- Internal parameters: $X_I = \theta = (\theta_1, \theta_2, \theta_3)$. Three prices pin down three parameters, so the system is square.
- Model functions: $V_i(\theta) = \text{Black}\big(F, K_i, T, \sigma(K_i;\theta)\big)$. No discounting is needed because the prices are un-discounted.

**Finding the unique $\theta$ (two-stage approach)**

*Stage 1 – implied vols.* Invert each price to a market implied vol $\hat\sigma_i$ by 1-D Newton:

$$\sigma^{(k+1)} = \sigma^{(k)} - \frac{\text{Black}(F,K_i,T,\sigma^{(k)}) - P_i}{\mathcal{V}_i(\sigma^{(k)})},$$

where $\mathcal{V}_i$ is Black vega. Since $\mathcal{V}_i > 0$, the Black price is strictly increasing in $\sigma$, so each $\hat\sigma_i$ is unique.

*Stage 2 – interpolator parameters.* Solve the $3\times3$ system

$$G(\theta) = \big(\sigma(K_i;\theta) - \hat\sigma_i\big)_{i=1,2,3} = 0$$

with the Newton recipe from Q1, using the Jacobian $J^\sigma_{ij} = \partial\sigma(K_i;\theta)/\partial\theta_j$.

(Equivalently, solve directly in price space with the Jacobian $J^P = \operatorname{diag}(\mathcal{V}_1,\mathcal{V}_2,\mathcal{V}_3)\,J^\sigma$, by the chain rule through the Black formula.)

*Example and uniqueness.* With log-moneyness $x = \ln(K/F)$, take

$$\sigma(K;\theta) = \theta_1 + \theta_2\,x + \theta_3\,x^2 .$$

This is linear in $\theta$, so Stage 2 is a single linear solve with a Vandermonde matrix in $(x_1, x_2, x_3)$. That matrix is invertible whenever the three strikes are distinct, so $\theta$ exists and is **unique**. Since $K_{atm} = F$ gives $x_2 = 0$, we get $\theta_1 = \hat\sigma_{atm}$ immediately. (In practice we also check that the resulting smile stays positive and arbitrage-free outside $[K_1, K_3]$.)

**How $\sigma(K)$ changes due to $\theta$**
- *Explicitly:* $\dfrac{\partial\sigma(K;\theta)}{\partial\theta}$ is just the derivative of the interpolator, available at any strike $K$. For the quadratic example it is $(1,\; x,\; x^2)$.
- *Through the market quotes* (by Q2):

$$\frac{d\theta}{dP} = \big(J^P\big)^{-1},\qquad
\frac{d\sigma(K)}{dP} = \frac{\partial\sigma(K;\theta)}{\partial\theta}\,\big(J^P\big)^{-1},\qquad
\frac{d\sigma(K)}{d\hat\sigma} = \frac{\partial\sigma(K;\theta)}{\partial\theta}\,\big(J^\sigma\big)^{-1}.$$

  For the quadratic example, $d\sigma(K)/d\hat\sigma_i = \ell_i(x)$, the Lagrange basis polynomial through the three strikes. The weights sum to 1, so a parallel shift in all three market vols shifts the entire smile in parallel.

**What $\theta$ might correspond to (intuition)**

Three points determine the **level, slope and curvature** of the smile, which match the three standard trader quotes:

| Parameter | Shape of the smile | Trader quote | Driven mainly by |
|---|---|---|---|
| $\theta_1$ | **Level**: shifts the smile up/down | ATM vol | $P_2$ |
| $\theta_2$ | **Slope / skew**: tilts the smile | Risk reversal ($\sigma_{K_3} - \sigma_{K_1}$) | $P_3$ vs $P_1$ |
| $\theta_3$ | **Curvature / smile**: lifts the wings vs ATM | Butterfly ($\tfrac12(\sigma_{K_1}+\sigma_{K_3}) - \sigma_{atm}$) | $P_1, P_3$ vs $P_2$ |

Parametrising $\theta$ this way, or simply choosing $\theta = (\sigma_{ATM}, RR, BF)$, makes $\partial\sigma(K)/\partial\theta$ read directly as vega exposure to ATM, risk reversal and butterfly, which are the instruments a trader actually hedges with. This is the same intuition behind SABR, whose parameters are read as $\alpha \approx$ level, $\rho \approx$ skew and $\nu \approx$ curvature.
